```python
# We'll first import some langchain dependencies to work with the model

from langchain_ollama import OllamaLLM

# Then we connect to the model

llm = OllamaLLM(
    model="llama3.2",
    temperature=0.3, 
    max_tokens=256,
    top_p=0.9,
    top_k=40,
    n_probs=2
)

# Then ask a question and get a response

response = llm.invoke("Why is the sky blue?")

print(response)
```
